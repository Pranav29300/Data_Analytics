# **Project Summary -**

This project is a comprehensive Exploratory Data Analysis (EDA) focused on classifying and predicting credit scores for Paisabazaar, a leading financial services company. The primary objective is to assist the organization in assessing customer creditworthiness, which is essential for risk management and providing personalized financial advice.

**Business Objective and Problem Statement**

- The core business goal is to develop a data-driven model that predicts individual credit scores based on various features such as income, credit card usage, and payment behavior. The analysis was guided by several key hypotheses:

- Determining if higher annual income correlates with better credit scores.

- Comparing the impact of the Credit Utilization Ratio (CUR) versus the total number of loans on credit scores.

- Evaluating whether a higher number of credit cards necessarily indicates a better score.

- Investigating the negative correlation between delayed payments and credit health.

**Dataset Overview**

- The project utilizes a substantial dataset comprising 100,000 rows and 28 columns. Key features include:

- Demographics: Age, Occupation, and SSN.

- Financial Metrics: Annual Income, Monthly Inhand Salary, and Outstanding Debt.

- Credit Details: Number of Bank Accounts, Number of Credit Cards, Credit History Age, and Credit Mix.

- Behavioral Data: Payment Behaviour, Amount Invested Monthly, and Credit Utilization Ratio.

- Target Variable: Credit Score (categorized as Good, Standard, or Poor).

**Analysis and Key Insights**

- The analysis followed the UBM (Univariate, Bivariate, and Multivariate) rule to extract meaningful patterns. Several of the initial assumptions were tested and either confirmed or refuted:

- Income and Credit Score: The analysis confirmed that a higher annual income generally correlates with a higher credit score.

- Loans vs. Credit Utilization: Contrary to the initial assumption, the number of loans taken was found to have a greater (negative) impact on credit scores than the Credit Utilization Ratio, which showed only a weak positive correlation.

- Credit Cards: Interestingly, the data revealed that a higher number of credit cards is often a sign of a Poor credit score rather than a Good one. Customers with high credit scores tended to manage fewer cards more effectively.

- Delayed Payments: As expected, there is a clear negative correlation between the number of delayed payments and credit scores; increased delays consistently lead to lower scores.

- Seasonal Trends: The project identified an 8-month trend in credit score variations, suggesting certain seasonal patterns in how scores fluctuate throughout the year.

**Conclusion and Strategic Impact**

The findings from this EDA provide actionable insights for Paisabazaar’s stakeholders. By identifying that the number of loans and credit cards are stronger indicators of poor credit than simple utilization ratios, the company can refine its risk assessment algorithms. Furthermore, understanding the negative impact of delayed payments and the positive influence of steady income allows for more targeted financial counseling and product recommendations. This structured approach to data analysis not only validates business assumptions but also uncovers counter-intuitive trends that can significantly improve loan approval accuracy and reduce default risks.

4.  **Num_of_Delayed_Payment and Interest_Rate (0.57):** This shows a moderately strong positive correlation. As the number of delayed payments increases, the interest rate tends to increase. This is a common practice by lenders, where a history of delayed payments signals higher risk, leading to higher interest rates on new or existing credit.

5.  **Num_of_Loan and Interest_Rate (0.56):** Similar to the previous point, a higher number of loans is positively correlated with higher interest rates. This could be because individuals with many loans might be seen as having a higher credit risk, resulting in higher interest rates.
